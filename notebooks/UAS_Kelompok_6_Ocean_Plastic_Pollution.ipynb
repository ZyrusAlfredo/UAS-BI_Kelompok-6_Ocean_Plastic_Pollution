# **Feature Engineering**

## **Memuat Data**

```python
dss_df = pd.read_csv('/content/integrated_ocean_plastic_dataset.csv')
```

## **Filter Data 2019**

```python
dss_df = merged_df[merged_df['year'] == 2019].copy()
```

```python
dss_df.shape
```

```python
dss_df.head()
```

## **Missing Value DSS Dataset**

```python
dss_df.isnull().sum()
```

### **Handling Missing Values DSS**

```python
dss_columns = [
    'ocean_pollution_share',
    'mismanaged_share',
    'recycled_share',
    'mismanaged_per_capita'
]

for col in dss_columns:
    dss_df[col] = dss_df[col].fillna(
        dss_df[col].median()
    )
```

## **Normalisasi Data**

### **Import Library**

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
features = [
    'ocean_pollution_share',
    'mismanaged_share',
    'recycled_share',
    'mismanaged_per_capita'
]
```

```python
scaler = MinMaxScaler()

dss_df[features] = scaler.fit_transform(
    dss_df[features]
)
```

## **Membuat Risk Score**

### **Implementasi Python**

```python
dss_df['risk_score'] = (
    (0.4 * dss_df['ocean_pollution_share']) +
    (0.3 * dss_df['mismanaged_share']) +
    (0.2 * dss_df['mismanaged_per_capita']) -
    (0.1 * dss_df['recycled_share'])
)
```

## **Ubah Ke Skala 0-100**

```python
dss_df['risk_score'] = dss_df['risk_score'] * 100
```

## **Membuat Risk Category**

```python
def categorize_risk(score):
    if score >= 80:
        return 'Critical'
    elif score >= 60:
        return 'High'
    elif score >= 40:
        return 'Medium'
    else:
        return 'Low'

dss_df['risk_category'] = dss_df['risk_score'].apply(categorize_risk)
```

## **Cek Hasil DSS**

```python
dss_df[['entity', 'risk_score', 'risk_category']].head(20)
```

## **Top High Risk Countries**

```python
dss_df.sort_values(
    by='risk_score',
    ascending=False
)[['entity', 'risk_score', 'risk_category']].head(10)
```

```python
dss_df.to_csv(
    'ocean_plastic_dss_dataset.csv',
    index=False
)
```

# **SQL Analysis**

Tahap SQL Analysis dilakukan untuk memperoleh insight dari Data Warehouse yang telah dibangun. Analisis dilakukan dengan menggabungkan fact table dan dimension table menggunakan query SQL. Hasil query digunakan untuk mengidentifikasi negara atau wilayah dengan tingkat risiko tertinggi, kontribusi polusi plastik laut terbesar, tingkat sampah tidak terkelola per kapita tertinggi, serta indikator pengelolaan sampah plastik seperti recycling, incineration, mismanaged waste, dan landfill.

SQL Analysis ini menjadi dasar dalam penyusunan dashboard Business Intelligence dan Decision Support System karena menghasilkan informasi yang dapat digunakan untuk monitoring, perbandingan antar wilayah, dan penentuan prioritas mitigasi pencemaran plastik laut.

## **Load Data Warehouse Ke SQLite**

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('ocean_plastic_dw.db')

dim_country.to_sql('dim_country', conn, if_exists='replace', index=False)
dim_date.to_sql('dim_date', conn, if_exists='replace', index=False)
dim_risk.to_sql('dim_risk', conn, if_exists='replace', index=False)
fact_table.to_sql('fact_ocean_pollution', conn, if_exists='replace', index=False)
```

## **Cek Tabel di Database**

```python
query = """
SELECT name
FROM sqlite_master
WHERE type='table';
"""

pd.read_sql_query(query, conn)
```

## **Query 1 - Melihat Data Gabungan**

```python
query = """
SELECT
    c.entity,
    c.code,
    d.year,
    f.plastic_production,
    f.ocean_pollution_share,
    f.recycled_share,
    f.mismanaged_share,
    f.landfilled_share,
    f.mismanaged_per_capita,
    f.risk_score
FROM fact_ocean_pollution f
JOIN dim_country c ON f.country_id = c.country_id
JOIN dim_date d ON f.date_id = d.date_id;
"""

pd.read_sql_query(query, conn)
```

## **Query 2 — Top 10 Negara/Wilayah dengan Risk Score Tertinggi**

```python
query = """
SELECT
    c.entity,
    d.year,
    f.risk_score
FROM fact_ocean_pollution f
JOIN dim_country c ON f.country_id = c.country_id
JOIN dim_date d ON f.date_id = d.date_id
ORDER BY f.risk_score DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

## **Query 3 — Top 10 Kontributor Polusi Plastik Laut**

```python
query = """
SELECT
    c.entity,
    d.year,
    f.ocean_pollution_share
FROM fact_ocean_pollution f
JOIN dim_country c ON f.country_id = c.country_id
JOIN dim_date d ON f.date_id = d.date_id
ORDER BY f.ocean_pollution_share DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

## **Query 4 — Negara/Wilayah dengan Mismanaged Waste Per Capita Tertinggi**

```python
query = """
SELECT
    c.entity,
    d.year,
    f.mismanaged_per_capita
FROM fact_ocean_pollution f
JOIN dim_country c ON f.country_id = c.country_id
JOIN dim_date d ON f.date_id = d.date_id
ORDER BY f.mismanaged_per_capita DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

## **Query 5 — Rata-Rata Waste Management Indicator**

```python
query = """
SELECT
    AVG(recycled_share) AS avg_recycled_share,
    AVG(incinerated_share) AS avg_incinerated_share,
    AVG(mismanaged_share) AS avg_mismanaged_share,
    AVG(landfilled_share) AS avg_landfilled_share
FROM fact_ocean_pollution;
"""

pd.read_sql_query(query, conn)
```

## **Query 6 — Analisis Produksi Plastik per Tahun**

```python
query = """
SELECT
    d.year,
    SUM(f.plastic_production) AS total_plastic_production
FROM fact_ocean_pollution f
JOIN dim_date d ON f.date_id = d.date_id
GROUP BY d.year
ORDER BY d.year;
"""

pd.read_sql_query(query, conn)
```

## **Query 7 — Negara dengan Recycling Rate Tertinggi**

```python
query = """
SELECT
    c.entity,
    d.year,
    f.recycled_share
FROM fact_ocean_pollution f
JOIN dim_country c ON f.country_id = c.country_id
JOIN dim_date d ON f.date_id = d.date_id
ORDER BY f.recycled_share DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

## **Query 8 — Negara Prioritas Mitigasi**

```python
query = """
SELECT
    c.entity,
    d.year,
    f.ocean_pollution_share,
    f.mismanaged_share,
    f.mismanaged_per_capita,
    f.recycled_share,
    f.risk_score
FROM fact_ocean_pollution f
JOIN dim_country c ON f.country_id = c.country_id
JOIN dim_date d ON f.date_id = d.date_id
WHERE f.risk_score >= 60
ORDER BY f.risk_score DESC;
"""

pd.read_sql_query(query, conn)
```
